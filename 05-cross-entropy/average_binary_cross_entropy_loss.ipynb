Exercise 3: Average Binary Cross-Entropy Loss across many examples (the actual loss function)

Goal: Compute loss across an entire dataset, the same way it's used in real training.

#### The setup
You have predictions from 3 different "models" on a small spam dataset:

y_true = np.array([1, 0, 1, 1, 0, 1, 0, 0, 1, 0])

##### Model A: random — always predicts 0.5
y_pred_A = np.full(10, 0.5)

##### Model B: pretty good
y_pred_B = np.array([0.9, 0.1, 0.85, 0.95, 0.2, 0.8, 0.15, 0.1, 0.92, 0.25])

##### Model C: confident but often wrong
y_pred_C = np.array([0.1, 0.9, 0.2, 0.05, 0.85, 0.15, 0.95, 0.8, 0.05, 0.9])

##### Your task
1. Implement binary_cross_entropy(y_true, y_pred) for the full batch
2. Compute average loss for each model
3. Compute accuracy for each model (with threshold 0.5)
4. Compare loss vs accuracy — what does loss tell you that accuracy doesn't?


```python
import numpy as np

y_true = np.array([1, 0, 1, 1, 0, 1, 0, 0, 1, 0])
```

```python
def binary_cross_entropy(y_true, y_pred):
    return -np.mean(y_true * np.log(y_pred) + (1-y_true) * np.log(1-y_pred))
```

```python
def accuracy(y_true, y_pred, thresold=0.5):
    prediction = (y_pred > thresold).astype(int)
    return np.mean(prediction == y_true)
```

```python
models = {
    'A (random, always 0.5)': np.full(10, 0.5),
    'B (good)':                np.array([0.9, 0.1, 0.85, 0.95, 0.2, 0.8, 0.15, 0.1, 0.92, 0.25]),
    'C (confident but wrong)': np.array([0.1, 0.9, 0.2, 0.05, 0.85, 0.15, 0.95, 0.8, 0.05, 0.9])
}
```

```python
print(f"{'Model':<35} {'Loss':<10} {'Accuracy':<10}")
print("-" * 55)
for name, y_pred in models.items():
    loss = binary_cross_entropy(y_true, y_pred)
    acc = accuracy(y_true, y_pred)
    print(f"{name:<35} {loss:<10.4f} {acc:<10.2%}")
```
