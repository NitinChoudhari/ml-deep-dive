```python
import numpy as np
```

```python
entropies = []
```

```python
# calculate entropy function
def calc_entropy(male, female, total = 100):
    p_male = male/total
    p_female = female/total

    entropy = p_male * np.log2(1/(p_male)) + p_female * np.log2(1/(p_female))
    entropies.append(entropy)
    
    print(f"Male={male}, Female={female} → Entropy = {entropy:.4f}")

max_entropy = np.max(entropies)
min_entropy = np.min(entropies)
```

#### np.argmax() — returns the index of the maximum value
#### np.argmin() — returns the index of the minimum value

```python
print(f"Max Entropy: {max_entropy:.4f}")
print(f"Min Entropy: {min_entropy:.4f}")
```

```python
calc_entropy(50,50)
calc_entropy(10,90)
calc_entropy(35,65)
calc_entropy(75,25)
calc_entropy(80,20)
# calc_entropy(0,20)
```
