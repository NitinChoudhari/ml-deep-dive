```python
import numpy as np
import matplotlib.pyplot as plt

# 20 houses, 3 features each: [size_sqft, bedrooms, age_years]
X = np.array([
    [450,  1, 15],
    [600,  1, 12],
    [750,  2, 20],
    [800,  2, 8],
    [900,  2, 25],
    [1000, 2, 5],
    [1100, 3, 18],
    [1200, 3, 10],
    [1250, 3, 30],
    [1400, 3, 7],
    [1500, 3, 15],
    [1600, 3, 3],
    [1750, 4, 22],
    [1800, 4, 12],
    [1900, 4, 28],
    [2000, 4, 5],
    [2200, 4, 18],
    [2400, 5, 10],
    [2600, 5, 8],
    [3000, 5, 15]
])

# Price in lakhs (₹)
y = np.array([
    32, 45, 50, 70, 55, 82, 75, 98, 80, 108,
    115, 135, 128, 148, 130, 170, 165, 200, 225, 270
])
```

```python
# Initialization

w = np.zeros(3)
b = 0.0

# hyperparameters
epochs = 5000
learning_rate = 0.01

X_scaled = X.astype(float).copy()
X_scaled[:, 0] = X_scaled[:,0] / 3000
X_scaled[:, 1] = X_scaled[:,1] / 5
X_scaled[:, 2] = X_scaled[:,2] / 30

n = len(X)
```

```python
for epoch in range(epochs):
     # make prediction
    y_pred = X_scaled @ w + b

    # compute error
    error = y_pred - y

    # compute the gradient i.e. calculus - chain rule
    dw = (2/n) * X_scaled.T @ error
    db = (2/n) * np.sum(error)

    # calc w and b
    w = w - learning_rate * dw
    b = b - learning_rate * db

    if epoch % 500 == 0:
        mse = np.mean(error**2)
        print(f"Epoch {epoch}: w = {w}, b = {b:.4f}, MSE = {mse:.4f}")
print(f"\nFinal: Epoch = {epoch}, w = {w}, b = {b:.4f}, MSE = {mse:.4f}")
```

```python
# Use it to predict
param = np.array([1500/3000, 3/5, 10/30])
prediction = param @ w + b
print(f"\nPredicted price: ₹{prediction:.2f} lakhs")
```
