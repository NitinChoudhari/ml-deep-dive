#### Linear regression draws the best straight line through a bunch of points so you can predict new values.
#### That's it. Everything else is detail.

#### Let's start with a real example

#### Imagine you have data on chickens at your farm:
#### Weight (kg)                 Price (₹)
####  1.0                           200
####  1.5                           280
####  2.0                           360
####  2.5                           450
####  3.0                           530

#### If I show you a chicken weighing 2.2 kg, what's the price?

#### Your brain already does linear regression. You see the pattern: heavier chicken = higher price, roughly ₹100 more per kg. So 2.2 kg ≈ ₹400.

#### Linear regression is just teaching a computer to do this automatically.

```python
import numpy as np

# Step 1: Our chicken data
X = np.array([1.0, 1.5, 2.0, 2.5, 3.0])    # weights
y = np.array([200, 280, 360, 450, 530])    # prices
```

```python
# Step 2: Initialize w and b randomly (start with zeros)
w = 0.0
b = 0.0
```

```python
# Step 3: Hyperparameters
learning_rate = 0.01 # how big a step to take each iteration
epochs = 1000 # how many times to update w and b

n = len(X) # number of data points
```

```python
for epoch in range(epochs):
    # Make predictions for the current w and b
    y_pred = w * X + b # this is just y = w*x + b for ALL points at once

    # compute the error
    error = y_pred - y # how far off are we?

    # Compute the gradients (this part comes from calculus — chain rule)
    # Don't memorize, understand: gradient tells us which direction to nudge w and b
    dw = (2/n) * np.sum(error * X) # gradient w.r.t. w
    db = (2/n) * np.sum(error) # gradient w.r.t. b

    #  update the w and b in the direction that reduces loss
    w = w - learning_rate * dw
    b = b - learning_rate * db

    # print progress every 100 epochs
    if epoch % 100 == 0:
        mse = np.mean(error**2)
        print(f"Epoch {epoch}: w = {w}, b = {b}, MSE = {mse}")
print(f"\nFinal: y = {w}*X + {b}")
```

```python
# Use it to predict
weight = 2.2
price = w * weight + b
print(f"A chicken weighting {weight} kg costs approximately {price}")
```

The 3 things you must understand
If you only remember 3 things, remember these:
1. The model is just a line: y = w*x + b The whole job is to find good w and b.
2. The loss tells us how bad the current line is We use MSE: average of squared errors. Lower = better.
3. Gradient descent adjusts w and b to reduce the loss
   Compute gradients (which way to nudge)
   Update parameters
   Repeat

That's it. Linear regression is literally these 3 ideas in a loop.
