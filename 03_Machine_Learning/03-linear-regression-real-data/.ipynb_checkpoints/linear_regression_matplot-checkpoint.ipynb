```python
import numpy as np
import matplotlib.pyplot as plt

# House size in sq ft
X = np.array([
    450, 600, 750, 800, 900, 1000, 1100, 1200, 1250, 1400,
    1500, 1600, 1750, 1800, 1900, 2000, 2200, 2400, 2600, 3000
])

# Price in lakhs (₹)
y = np.array([
    32, 45, 50, 65, 60, 78, 72, 95, 88, 102,
    115, 125, 130, 145, 138, 165, 175, 195, 220, 270
])
```

```python
plt.scatter(X, y, color='blue', label='Actual data')
plt.xlabel('Size (sq ft)')
plt.ylabel('Price (₹ lakhs)')
plt.title('Pune House Prices')
plt.show()
```

```python
# Initialization

w = 0.0
b = 0.0

# hyperparameters
epochs = 1000
learning_rate = 0.01

X_scaled = X/1000

n = len(X)
```

```python
for epoch in range(epochs):
     # make prediction
    y_pred = w * X_scaled + b

    # compute error
    error = y_pred - y

    # compute the gradient i.e. calculus - chain rule
    dw = (2/n) * np.sum(error * X_scaled)
    db = (2/n) * np.sum(error)

    # calc w and b
    w = w - learning_rate * dw
    b = b - learning_rate * db

    if epoch % 100 == 0:
        mse = np.mean(error**2)
        print(f"Epoch {epoch}: w = {w:.4f}, b = {b:.4f}, MSE = {mse:.4f}")
print(f"\nFinal: Epoch = {epoch}, w = {w:.4f}, b = {b:.4f}, MSE = {mse:.4f}")
```

```python
# Use it to predict
param = 1350
prediction = w * param + b
print(f"The house of {param} sqft costs approximately {prediction:.2f}")
```

```python
plt.scatter(X, y, color='blue', label='Actual')
plt.plot(X, w*X_scaled + b, color='red', label='My model')
plt.xlabel('Size (sq ft)')
plt.ylabel('Price (₹ lakhs)')
plt.legend()
plt.show()
```

```python
from sklearn.linear_model import LinearRegression

sk_model = LinearRegression()
sk_model.fit(X_scaled.reshape(-1, 1), y)

print(f"My implementation:")
print(f"  w = {w:.4f}, b = {b:.4f}")
print(f"\nsklearn:")
print(f"  w = {sk_model.coef_[0]:.4f}, b = {sk_model.intercept_:.4f}")
```
